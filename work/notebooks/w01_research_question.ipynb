## 2. The question: decision, action, cost of a wrong call

**Search question:** Among pages with enough search exposure, which should an editor review first for a possible click-through or engagement opportunity, after comparing CTR within similar average-position bands? **Unit of analysis:** one pseudonymized content page, summarized over the starter dataset's trailing 90-day window. **Output:** a ranked review queue with evidence-based reason codes (for example, enough impressions, position band, and below-band CTR); it is a shortlist for investigation, not an automatic edit instruction. **Action:** an editor checks the search snippet and intent match, then may revise title/meta copy, inspect the page experience, or monitor it. **Cost of a wrong call:** a false positive wastes limited editorial time and could prompt an unnecessary change; a false negative leaves a plausible opportunity unreviewed. I will favor reviewable evidence and avoid recommending changes on low-volume pages. Success should be judged at the team's review capacity (such as precision@K against a clearly defined observed outcome), not generic accuracy.

Data or ML helps because position, exposure, intent, and engagement interact across many pages; however, if a simple position-adjusted rule works as well, that is the better answer.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

DATA_URL = 'https://raw.githubusercontent.com/Ashritha-boop/fly_machine/main/data/raw/content_refresh_anonymized.csv'
data = pd.read_csv(DATA_URL)

eligible = data[(data['impressions_90d'] >= 500) & (data['avg_position'] > 0) & (data['avg_position'] <= 20)]
candidates = eligible[eligible['ctr'] < 0.5]

print(f'Starter data: {len(data):,} pages across {data["client_id"].nunique()} pseudonymized clients')
print(f'Pages with >=500 impressions and average position 1-20: {len(eligible):,}')
print(f'Of those, pages with CTR <0.5%: {len(candidates):,}')
print(f'Median impressions: eligible={eligible["impressions_90d"].median():,.0f}; candidates={candidates["impressions_90d"].median():,.0f}')
print(eligible.groupby('position_tier', observed=True)['ctr'].median().sort_index().round(2).to_string())
```

## 4. Careful words: what I can and can't claim

I can report **observed** differences in the anonymized starter snapshot and build **directional, decision-support** rankings for human review. CTR is stored as a percentage point value (for example, 0.5 means 0.5%), and it must be interpreted within position bands with adequate impression volume. The starter snapshot is observational and its CTR threshold identifies candidates, not pages proven to benefit from edits. It cannot show that a title change caused more clicks, that any page will recover, or that I have discovered a Google ranking factor or predicted Google's algorithm.
